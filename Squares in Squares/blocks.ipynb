### Squares in Squares Kaggle Challenge
The goal of this challenge is to fit $n$ squares in as tight a box as possible. First of all, we will try to implement a baseline, in which the blocks are simply stacked on top of eachother, with the same angle for each one.

```python
import pandas as pd
import numpy as np
```

```python
square_collection_list = []

for amount_of_squares in range(1, 1000):
    amount_of_rows_and_columns = np.ceil(np.sqrt(amount_of_squares))
    print(amount_of_rows_and_columns)
    for square_idx in range(amount_of_squares):

        square_collection_list.append([amount_of_squares, square_idx, 0.5 + (square_idx // amount_of_rows_and_columns), 0.5 + (square_idx % amount_of_rows_and_columns), 0])

placement_df = pd.DataFrame(square_collection_list, columns=["n", "square_id", "x", "y" ,"angle"])
placement_df["id"] = list(range(len(placement_df)))
```

```python
placement_df.to_csv("baseline_submission.csv")
```
